# 01 · Inspect the capture videos

Check that the three videos load, have the expected resolution and orientation, and look at frames spread over each one.

```python
import matplotlib.pyplot as plt

from avatar_splats.config import VIDEOS, load_config
from avatar_splats.data import probe, read_frames

config = load_config("../configs/default.yaml")
videos = {name: getattr(config.capture, name) for name in VIDEOS}
```

```python
infos = {name: probe(path) for name, path in videos.items()}
for name, info in infos.items():
    print(f"{name:<18} {info.width}x{info.height}  {info.fps:.2f} fps  {info.num_frames} frames")
```

Eight frames spread evenly over each video. The performer should be upright: a sideways frame means the file carries rotation metadata that the decoder ignores.

```python
columns = 8
fig, axes = plt.subplots(len(videos), columns, figsize=(2 * columns, 5 * len(videos)))
for row, (name, path) in zip(axes, videos.items(), strict=True):
    step = max(1, infos[name].num_frames // columns)
    for ax, frame in zip(row, read_frames(path, step=step, max_frames=columns), strict=False):
        ax.imshow(frame)
        ax.axis("off")
    row[0].set_title(name, loc="left")
fig.tight_layout()
```
